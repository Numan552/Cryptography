```python
print("===== Hill Cipher (2x2) =====")

text = input("Enter Plain Text: ").upper()

print("Enter 2x2 Key Matrix:")

key = []

for i in range(2):
    row = list(map(int, input().split()))
    key.append(row)

if len(text) % 2 != 0:
    text += "X"

cipher = ""

for i in range(0, len(text), 2):

    p1 = ord(text[i]) - 65
    p2 = ord(text[i + 1]) - 65

    c1 = (key[0][0] * p1 + key[0][1] * p2) % 26
    c2 = (key[1][0] * p1 + key[1][1] * p2) % 26

    cipher += chr(c1 + 65)
    cipher += chr(c2 + 65)

print("\nCipher Text:", cipher)
```
