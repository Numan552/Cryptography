```python
print("===== XOR Block Cipher =====")
print("1. Encrypt")
print("2. Decrypt")

choice = int(input("Enter your choice: "))

text = input("Enter the text: ")
key = input("Enter the key: ")

result = []

for i in range(len(text)):
    x = ord(text[i]) ^ ord(key[i % len(key)])

    if choice == 1:
        result.append(format(x, "02X"))
    else:
        result.append(chr(x))

if choice == 1:
    print("\nEncrypted (Hex):")
    print(" ".join(result))
else:
    print("\nDecrypted Text:")
    print("".join(result))
```
