```python
def shift_cipher(text, key):

    result = ""

    for ch in text:

        if ch.isupper():
            result += chr((ord(ch)-65+key)%26+65)

        elif ch.islower():
            result += chr((ord(ch)-97+key)%26+97)

        else:
            result += ch

    return result


print("===== Shift (Caesar) Cipher =====")
print("1. Encrypt")
print("2. Decrypt")

choice = int(input("Enter your choice: "))

text = input("Enter the text: ")

key = int(input("Enter the key (0-25): "))

if choice == 2:
    key = 26 - key

result = shift_cipher(text, key)

if choice == 1:
    print("\nEncrypted Text:", result)
else:
    print("\nDecrypted Text:", result)
```
