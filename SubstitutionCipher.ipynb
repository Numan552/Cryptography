```python
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

print("===== Substitution Cipher =====")
print("1. Encrypt")
print("2. Decrypt")

choice = int(input("Enter your choice: "))

text = input("Enter the text: ")
key = input("Enter 26-letter substitution key: ").upper()

result = ""

for ch in text:

    if ch.isupper():

        if choice == 1:
            result += key[ord(ch)-65]

        else:
            result += alphabet[key.index(ch)]

    elif ch.islower():

        upper = ch.upper()

        if choice == 1:
            result += key[ord(upper)-65].lower()

        else:
            result += alphabet[key.index(upper)].lower()

    else:
        result += ch

if choice == 1:
    print("\nEncrypted Text:", result)
else:
    print("\nDecrypted Text:", result)
```
